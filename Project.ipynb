```python
# Import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Original data
Original_data = pd.read_csv("Dataset.csv")
```

```python
Original_data
```

```python
# New data
New_Data = Original_data[
    (Original_data["Quantity"] > 0) &
    (Original_data["Price"] > 0)
].copy()
```

```python
New_Data.head(10)
```

#                                                   Data cleaning

```python
New_Data.info()
```

```python
## Data type
New_Data['InvoiceDate'] = pd.to_datetime(New_Data['InvoiceDate'])
New_Data["Customer ID"] = New_Data["Customer ID"].astype(int)
New_Data["Customer ID"] = New_Data["Customer ID"].astype(int)
```

```python
# Missing values
New_Data.isnull().sum()
```

```python
New_Data.dropna(subset = ["Customer ID"] , inplace = True)  
```

```python
New_Data.duplicated().sum()
```

```python
New_Data["Revenue"] = New_Data["Price"] * New_Data["Quantity"]
```

# Calculating RFM - 

# Recency

```python
Reference_date = pd.to_datetime(New_Data["InvoiceDate"].max())
print(Reference_date)
```

```python
New_Data["Recency"] = (
    Reference_date - New_Data.groupby("Customer ID")["InvoiceDate"].transform("max")
).dt.days
```

```python
New_Data
```

# Frequency

```python
Frequency = New_Data.groupby("Customer ID")["Invoice"].nunique()
```

```python
New_Data["Frequency"] = New_Data["Customer ID"].map(Frequency)
```

# Monitary

```python
Total_spending = New_Data.groupby("Customer ID")["Revenue"].sum()
```

```python
Total_spending
```

```python
New_Data["Monitary"] = New_Data["Customer ID"].map(Total_spending)
```

```python
New_Data.head(10)
```

```python
New_Data["Recency"].describe()
```

```python
def recency_score(days):
    if days <= 21:
        return 5
    elif days <= 76:
        return 4
    elif days <= 150:
        return 3
    elif days <= 300:
        return 2
    else:
        return 1
New_Data["Recency_Score"] = New_Data["Recency"].apply(recency_score)
```

```python
New_Data["Frequency"].describe()
```

```python
def frequency_score(count):
    if count <= 5:
        return 1
    elif count <= 12:
        return 2
    elif count <= 26:
        return 3
    elif count <= 50:
        return 4
    else:
        return 5

New_Data["Frequency_Score"] = New_Data["Frequency"].apply(frequency_score)
```

```python
New_Data["Monitary"].describe()
```

```python
def monetary_score(amount):
    if amount <= 1771.45:
        return 1
    elif amount <= 4525.46:
        return 2
    elif amount <= 11403.41:
        return 3
    elif amount <= 30000:
        return 4
    else:
        return 5

New_Data["Monetary_Score"] = New_Data["Monitary"].apply(monetary_score)
```

```python
def customer_Segment(Score):
    r = Score["Recency_Score"]
    f = Score["Frequency_Score"]
    m = Score["Monetary_Score"]
    
    if r >= 4 and f >= 4 and m >= 4:
        return "Premium"
    elif r >= 3 and f == 5:
        return "Loyal"
    elif r >= 4 and f < 3:
        return "Potential"
    elif r <= 2 and f >= 4 and m >= 4:
        return "At Risk"
    elif r <=2 and f <= 2 and m <= 2:
        return "In-Active"
    else:
        return "Regular"
New_Data["Customer segment"] = New_Data.apply(customer_Segment , axis = 1)
```
